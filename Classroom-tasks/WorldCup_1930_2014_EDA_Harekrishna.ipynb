16. Compare average attendance across years for each stage.

17. Compare Home Goals and Away Goals across tournament stages.

```python
stage_goals = df.groupby('Stage')[['Home Team Goals', 'Away Team Goals']].mean()
stage_goals.plot(kind='bar', figsize=(15,7))
plt.title('Average Home and Away Goals by Tournament Stage')
plt.xlabel('Tournament Stage')
plt.ylabel('Average Goals')
plt.xticks(rotation=90)
plt.legend(['Home Goals', 'Away Goals'])
plt.show()
```

Ans:
1. The average Home Goals are higher than the average Away Goals in every tournament stages.
2. The Preliminary Round has the highest average Home Goals, while the Play-off for Third Place has the highest average Away Goals.
3. There is small difference between Home Goals and Away Goals in some grop from A to Group H but larger difference in First Round, Group 6 and Preliminary round.

18. Is there any correlation among Attendance, Home Goals, Away Goals?

```python
# Caluculating correlation
corr = df[['Attendance', 'Home Team Goals', 'Away Team Goals']].corr()
corr
```

```python
# plotting in heat map
plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation among Attendance, Home Goals and Away Goals')
plt.show()
```

Ans:
1. There is no strong corelation between Attendance, Home Team Goal, and Away Team Goal.
2. There is only one positive corealtion between Away Team Goal and Home Team Goal which is 0.012 which is too less.
3. Attendance have very negativ corelation with Home Team Goal and also negative corelation with Away team goal by -0.03 .

20. Pair Plot

```python
plt.figure(figsize=(18,15))
plt.suptitle(
    'FIFA World Cup (1930–2014) Data Analysis Dashboard',
    fontsize=20,
    fontweight='bold'
)


# Top 10 Highest Attendance Matches
plt.subplot(3,2,1)
top_att = df.nlargest(10, 'Attendance').copy()
top_att['Match'] = top_att['Home Team Name'] + ' vs ' + top_att['Away Team Name']
sns.barplot(
    data=top_att,
    x='Attendance',
    y='Match',
    errorbar=None,
    palette='viridis'
)
plt.title('Top 10 Highest Attendance Matches')
plt.xlabel('Attendance')
plt.ylabel('Match')

# Average Goals per Match by World Cup Year
plt.subplot(3,2,2)
avg_goal = df.groupby('Year')[['Home Team Goals','Away Team Goals']].mean()
avg_goal['Average Goals'] = (
    avg_goal['Home Team Goals']
    + avg_goal['Away Team Goals']
)
sns.lineplot(
    x=avg_goal.index,
    y=avg_goal['Average Goals'],
    marker='o',
    linewidth=2
)
plt.title('Average Goals per Match by Year')
plt.xlabel('World Cup Year')
plt.ylabel('Average Goals')
plt.grid(True)


# Attendance Distribution
plt.subplot(3,2,3)
sns.histplot(
    data=df,
    x='Attendance',
    bins=30,
    kde=True,
    color='skyblue'
)
plt.title('Distribution of Match Attendance')
plt.xlabel('Attendance')
plt.ylabel('Frequency')

# Home vs Away Goals Distribution
plt.subplot(3,2,4)
goal_df = df[['Home Team Goals','Away Team Goals']]
sns.boxplot(data=goal_df, palette='Set2')
plt.title('Distribution of Home and Away Goals')
plt.xlabel('')
plt.ylabel('Goals')


# 5. Matches Played in Each World Cup
plt.subplot(3,2,5)
match_year = df.groupby('Year').size().reset_index(name='Matches')
sns.barplot(
    data=match_year,
    x='Year',
    y='Matches',
    errorbar=None,
    palette='magma'
)
plt.title('Matches Played in Each World Cup')
plt.xlabel('Year')
plt.ylabel('Number of Matches')
plt.xticks(rotation=90)


# 6. Home Goals vs Away Goals
plt.subplot(3,2,6)
sns.scatterplot(
    data=df,
    x='Home Team Goals',
    y='Away Team Goals',
    alpha=0.7,
    color='red'
)
plt.title('Relationship Between Home and Away Goals')
plt.xlabel('Home Team Goals')
plt.ylabel('Away Team Goals')
plt.grid(True)



plt.tight_layout(rect=[0,0,1,0.95], pad=3)
plt.subplots_adjust(hspace=0.6, wspace=0.5)
plt.subplots_adjust(
    hspace=0.6,
    wspace=0.5
)
plt.show()
```
